## 6. Prioritization method

### 6.1 Factors and weights

Weights total **100%**. Each factor is scored from 1 to 5, where 5 is more attractive. For risk, the factor is **risk manageability**: a high score means risks are easier to control.

| Factor | Weight | Score 1 | Score 3 | Score 5 |
|---|---:|---|---|---|
| Business value | 30% | Weak/unclear link to stated goals | Useful but value is uncertain or indirect | Strong link to a material, measurable business problem |
| Data readiness | 20% | Essential data absent or unusable | Some data exists, but important gaps need work | Relevant data is available, approved, and sufficiently reliable |
| Feasibility | 20% | Difficult to implement within constraints | Achievable with notable integration/process work | Bounded implementation using accessible tools and workflows |
| Risk manageability | 15% | Major risks are difficult to control | Risks can be managed with substantial controls | Risks are well understood and controllable with practical safeguards |
| Time to value | 15% | Benefits unlikely to be assessed in 12 months | Useful evidence likely after meaningful setup | A measurable pilot can produce evidence quickly |

### 6.2 Scoring assumptions

Scores are reasoned judgments based on the supplied case, not measured facts. Value is scored on the strength of the business problem and measurability, not an invented monetary return. The customer-enquiry score assumes approved policy material can be assembled. Scores should be revisited after the first 30-day data and process review.

Weighted score formula:

\[
\text{Weighted score} = \sum_i \left(\frac{\text{weight}_i}{100} \times \text{score}_i\right)
\]

The maximum possible score is 5.00.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

weights = {
    "Business value": 30,
    "Data readiness": 20,
    "Feasibility": 20,
    "Risk manageability": 15,
    "Time to value": 15,
}
assert sum(weights.values()) == 100, "Weights must total 100%."

scores = {
    "Inventory forecasting": {
        "Business value": 5, "Data readiness": 3, "Feasibility": 3,
        "Risk manageability": 3, "Time to value": 3
    },
    "Customer-enquiry assistance": {
        "Business value": 4, "Data readiness": 4, "Feasibility": 4,
        "Risk manageability": 4, "Time to value": 5
    },
    "Marketing optimization": {
        "Business value": 4, "Data readiness": 3, "Feasibility": 3,
        "Risk manageability": 2, "Time to value": 3
    },
    "Computer-vision shelf inspection": {
        "Business value": 3, "Data readiness": 1, "Feasibility": 2,
        "Risk manageability": 3, "Time to value": 2
    },
}

def calculate_priority(score_dict, weight_dict):
    rows = []
    for use_case, factor_scores in score_dict.items():
        contributions = {
            factor: (weight_dict[factor] / 100) * factor_scores[factor]
            for factor in weight_dict
        }
        rows.append({
            "Use case": use_case,
            **{f"{factor} score": factor_scores[factor] for factor in weight_dict},
            **{f"{factor} contribution": round(value, 3) for factor, value in contributions.items()},
            "Weighted total": round(sum(contributions.values()), 3),
        })
    result = pd.DataFrame(rows).sort_values("Weighted total", ascending=False).reset_index(drop=True)
    result.insert(0, "Rank", range(1, len(result) + 1))
    return result

priority = calculate_priority(scores, weights)
# Display the ranking as a readable, presentation-ready table.
styled_priority = (
    priority[["Rank", "Use case", "Weighted total"]]
    .style
    .format({"Weighted total": "{:.2f} / 5"})
    .background_gradient(subset=["Weighted total"], cmap="GnBu")
    .set_properties(**{"text-align": "left", "padding": "8px"})
    .set_table_styles([
        {"selector": "th", "props": [("background-color", "#123b5d"), ("color", "white"),
                                    ("padding", "9px"), ("text-align", "left")]},
        {"selector": "caption", "props": [("caption-side", "top"), ("font-size", "16px"),
                                          ("font-weight", "bold"), ("color", "#123b5d")]}
    ])
    .set_caption("AI use-case priority ranking")
)
display(styled_priority)
print(f"Scoring weights total: {sum(weights.values())}%")

# Visual comparison of weighted scores.
plot_data = priority.sort_values("Weighted total", ascending=True)
fig, ax = plt.subplots(figsize=(10, 4.8))
bars = ax.barh(plot_data["Use case"], plot_data["Weighted total"], color=["#8ab8c8", "#8ab8c8", "#8ab8c8", "#168a8a"])
ax.set_xlim(0, 5)
ax.set_xlabel("Weighted score (out of 5)")
ax.set_title("Which AI opportunity should Horizon Retail pilot first?", loc="left", pad=16, weight="bold")
ax.grid(axis="x", linestyle="--", alpha=0.25)
ax.set_axisbelow(True)
for bar, value in zip(bars, plot_data["Weighted total"]):
    ax.text(value + 0.06, bar.get_y() + bar.get_height()/2, f"{value:.2f}", va="center", fontsize=10, weight="bold")
ax.spines[["top", "right", "left"]].set_visible(False)
ax.spines["bottom"].set_alpha(0.25)
plt.tight_layout()
plt.show()
````

### 6.3 Score rationale for all 20 ratings

|---|---|---|---|---|---|
| Inventory forecasting | **5** — stockouts and ₹45 lakh write-offs are material stated issues, although preventable value is unknown. | **3** — 18 months of records exist, but 15% of expected stock-count entries are missing. | **3** — forecasting is feasible in principle but needs data reconciliation and workflow integration. | **3** — bad recommendations could affect availability; staff review and conservative limits can reduce risk. | **3** — a pilot can be tested, but clean data and enough evaluation periods are needed. |
| Customer-enquiry assistance | **4** — high enquiry volume and 10-hour response delay; savings and customer impact still need measurement. | **4** — routine policy topics are identified, but actual policy-content quality must be checked. | **4** — a bounded assistant can draft answers for staff without replacing the service process. | **4** — retrieval, approved sources, human approval, and escalation make risks manageable. | **5** — response time and answer quality can be measured during a short supervised pilot. |
| Marketing optimization | **4** — potential to improve campaign performance, but incremental margin is not provided. | **3** — records and conversion figures exist, but consent is recorded for only 65%. | **3** — segmentation and experiments are feasible after data and consent cleanup. | **2** — consent, privacy, unwanted targeting, and misleading attribution require significant controls. | **3** — a properly designed campaign test can provide evidence, but preparation is necessary. |

## 7. Sensitivity analysis

### Alternative assumption tested

The base case assumes current approved policy information can be assembled, giving customer-enquiry assistance a data-readiness score of 4, feasibility 4, and risk-manageability 4. A consequential downside case is that policy documents are fragmented, out of date, or not approved for use. In that case, reduce customer-enquiry readiness from **4 to 2**, feasibility from **4 to 3**, and risk manageability from **4 to 3**. All weights and other scores remain unchanged.

This is a scenario test, not a claim that the policy content is actually poor. The first 30 days should resolve this uncertainty.

````python
sensitivity_scores = {k: v.copy() for k, v in scores.items()}
sensitivity_scores["Customer-enquiry assistance"]["Data readiness"] = 2
sensitivity_scores["Customer-enquiry assistance"]["Feasibility"] = 3
sensitivity_scores["Customer-enquiry assistance"]["Risk manageability"] = 3

base_result = calculate_priority(scores, weights)[["Use case", "Rank", "Weighted total"]].rename(
    columns={"Rank": "Base rank", "Weighted total": "Base score"}
)
sensitivity_result = calculate_priority(sensitivity_scores, weights)[["Use case", "Rank", "Weighted total"]].rename(
    columns={"Rank": "Sensitivity rank", "Weighted total": "Sensitivity score"}
)
sensitivity = base_result.merge(sensitivity_result, on="Use case")
sensitivity["Score change"] = (sensitivity["Sensitivity score"] - sensitivity["Base score"]).round(3)
display(
    sensitivity.sort_values("Sensitivity rank").style
    .format({"Base score": "{:.2f}", "Sensitivity score": "{:.2f}", "Score change": "{:+.2f}"})
    .background_gradient(subset=["Sensitivity score"], cmap="GnBu")
    .set_properties(**{"padding": "8px"})
    .set_table_styles([{"selector": "th", "props": [("background-color", "#123b5d"),
                                                     ("color", "white"), ("padding", "9px")]}])
    .set_caption("Base case vs. downside assumption")
)

# Make the impact of the changed assumption easy to compare.
comparison = sensitivity.sort_values("Base score", ascending=True)
y = np.arange(len(comparison))
fig, ax = plt.subplots(figsize=(10, 4.8))
ax.barh(y + 0.18, comparison["Base score"], height=0.34, label="Base case", color="#2b7a9b")
ax.barh(y - 0.18, comparison["Sensitivity score"], height=0.34, label="Downside case", color="#8bc6bd")
ax.set_yticks(y, comparison["Use case"])
ax.set_xlim(0, 5)
ax.set_xlabel("Weighted score (out of 5)")
ax.set_title("Does the recommendation change if policy data is not ready?", loc="left", pad=16, weight="bold")
ax.grid(axis="x", linestyle="--", alpha=0.25)
ax.set_axisbelow(True)
ax.legend(frameon=False, ncol=2, loc="lower right")
ax.spines[["top", "right", "left"]].set_visible(False)
ax.spines["bottom"].set_alpha(0.25)
plt.tight_layout()
plt.show()
````

### Sensitivity conclusion

Under the downside assumption, customer-enquiry assistance falls from its base score of 4.15 to 3.40. Inventory forecasting remains at 3.60 and becomes the highest-ranked option. Therefore, the recommendation is **conditional**: choose customer-enquiry assistance if policy readiness is confirmed during the first month; otherwise, reassess whether inventory is the better first pilot after data-quality checks. This prevents a ranking from being treated as more certain than the evidence supports.

## 8. Recommendation and alternatives deferred

### Recommended first pilot: customer-enquiry assistance

Start with a supervised assistant for a limited set of routine questions about delivery, returns, and product policies. It should retrieve information only from approved sources and draft responses for a trained staff member to review. Unknown, conflicting, sensitive, or exceptional cases must be escalated to a person. Keep the existing customer-service process available as a fallback.

**Why first:** the case gives a measurable volume (8,000 enquiries/month), a clear service baseline (10-hour average first response), and a bounded topic group (about 60% of enquiries). These make a 90-day test possible without assuming a guaranteed financial return.

### Why not the other options now?

1. **Inventory forecasting — defer, but retain as the fallback.** It addresses important problems, including 6% item unavailability and ₹45 lakh annual write-offs. However, 15% of expected stock-count entries are missing and the case does not establish preventable write-off value, lead times, or the share of stockouts attributable to forecasting. It becomes more attractive after reconciliation and baseline testing, or immediately if the customer-policy knowledge base fails the readiness gate.
2. **Marketing optimization — defer.** Only 65% of customer records have recorded marketing consent. Resolve consent and suppression records first, define incremental margin, and design a controlled experiment. Reconsider after these safeguards and data definitions are reliable.
3. **Computer-vision shelf inspection — defer.** There is no usable shelf-image dataset and inspection outcomes are inconsistent. Standardize inspection records and collect permitted, representative, labeled images before considering a model pilot.

A lower-cost non-AI improvement—such as searchable FAQs and approved answer templates—should be included as a baseline or comparison, not ignored in favor of AI for its own sake.

### Proposed acceptance and decision gates

These are suggested targets for approval before the pilot—not established facts:

- **Readiness:** all content in the pilot scope has a named owner, current version, and approval; privacy/security review complete; test set and baseline signed off.
- **Pre-live answer test:** at least 95% of test answers judged correct and supported by approved sources; 100% of known high-risk/uncertain test cases must trigger the defined escalation. If the test set is too small to interpret, extend testing.
- **Live oversight:** 100% of AI drafts are reviewed by staff before sending throughout the 90-day pilot.
- **Service outcome:** target at least a 30% reduction in median first-response time for in-scope enquiries compared with the agreed baseline, without deterioration in answer quality or customer satisfaction. Confirm that the baseline and denominator are comparable.
- **Safety:** zero unresolved critical privacy/security incidents; any critical incident triggers pause and review.
- **Decision:** **continue** if safety/quality gates pass and service benefit is credible; **revise** if quality is acceptable but workflow or outcome targets are missed; **stop** if critical safety issues remain, policy sources cannot be trusted, or the pilot shows no credible benefit after agreed adjustments.

## 13. Limitations and next steps
